<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> یکی بودن حجم برای اطمینان از سالم بودن فایل کافی نیست.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> فایل‌ها روی دیسک کولب کپی شدند و sha256 هر کدام حساب و با مقدار هاگینگ‌فیس مقایسه شد. هر دو فایل هم از نظر حجم و هم از نظر sha256 درست بودند. یعنی داده‌ی من بایت‌به‌بایت همانی است که بقیه‌ی گروه دارند.</p>
</div>

```python
def sha256_of(path, block=16 * 1024 * 1024):
    h = hashlib.sha256()
    with open(path, 'rb') as f:
        for chunk in iter(lambda: f.read(block), b''):
            h.update(chunk)
    return h.hexdigest()


manifest = {'repo': REPO, 'revision': REV, 'files': {}}
for path, meta in remote.items():
    src, dst = f'{DATA}/{path}', f'{LOCAL}/{os.path.basename(path)}'
    if not os.path.exists(dst) or os.path.getsize(dst) != meta['size']:
        shutil.copy(src, dst)
    digest = sha256_of(dst)
    ok_size = os.path.getsize(dst) == meta['size']
    ok_sha = meta['sha256'] is None or digest == meta['sha256']
    manifest['files'][path] = {'local': dst, 'bytes': meta['size'], 'sha256': digest,
                               'size_ok': ok_size, 'sha256_ok': ok_sha}
    print(f'{os.path.basename(path)}: size {"OK" if ok_size else "MISMATCH"} | sha256 {"OK" if ok_sha else "MISMATCH"}')
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">norm_series</span> این یکسان‌سازی را انجام می‌دهد و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">parse_list</span> نکات مثبت و منفی را از حالت لیست بیرون می‌آورد. <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">build_text</span> هم عنوان، متن، نکات مثبت و نکات منفی را با « | » کنار هم می‌گذارد. جلوی دو فیلد آخر برچسب «نکات مثبت:» و «نکات منفی:» می‌آید تا مدل بداند هر تکه از کجا آمده است.</p>
</div>

```python
AR_FIX = {'\u064a': '\u06cc', '\u0649': '\u06cc', '\u0643': '\u06a9', '\u06d5': '\u0647'}
AR_FIX.update({chr(0x660 + i): chr(0x6F0 + i) for i in range(10)})
AR_FIX = str.maketrans(AR_FIX)
DROP_RE = r'[\u064b-\u065f\u0670\u0640\u200d\u200e\u200f\u202a-\u202e\u2066-\u2069\ufeff]'
POS_TAG = '\u0646\u06a9\u0627\u062a \u0645\u062b\u0628\u062a: '
NEG_TAG = '\u0646\u06a9\u0627\u062a \u0645\u0646\u0641\u06cc: '
LIST_SEP = ' \u060c '


def norm_series(s):
    s = s.fillna('').astype(str).str.translate(AR_FIX)
    s = s.str.replace(DROP_RE, '', regex=True)
    s = s.str.replace(r'\u200c{2,}', '\u200c', regex=True)
    s = s.str.replace(r'[ \t]+\u200c[ \t]*|\u200c[ \t]+', ' ', regex=True)
    s = s.str.replace(r'([^\W\d_])\1{2,}', r'\1\1', regex=True)
    s = s.str.replace(r'([^\w\s])\1{3,}', r'\1\1\1', regex=True)
    s = s.str.replace(r'\s+', ' ', regex=True).str.strip()
    return s.str.strip('\u200c ')


def parse_list(v):
    v = v.strip()
    if v in ('', '[]', 'nan', 'None', 'null'):
        return ''
    if v.startswith('['):
        for f in (json.loads, ast.literal_eval):
            try:
                items = f(v)
                if isinstance(items, (list, tuple)):
                    return LIST_SEP.join(str(x).strip() for x in items if str(x).strip())
            except Exception:
                pass
        return v.strip('[]')
    return v


def parse_col(s):
    s = s.fillna('').astype(str)
    m = s.str.len() > 2
    s = s.where(m, '')
    s[m] = s[m].map(parse_list)
    return s


def build_text(title, body, adv, dis):
    return [' | '.join(p for p in (t, b, (POS_TAG + a) if a else '', (NEG_TAG + d) if d else '') if p)
            for t, b, a, d in zip(title, body, adv, dis)]
```

<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۸. خواندن کل فایل به صورت تکه‌تکه</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> فایل شش میلیون ردیفی را نمی‌شد یک‌جا در حافظه‌ی کولب پردازش کرد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> فایل در تکه‌های ۵۰۰ هزارتایی خوانده شد و برای هر تکه متن ورودی ساخته شد. فقط ستون‌های لازم ماندند: برچسب، شناسه‌ی محصول، امتیاز، سال، داشتن یا نداشتن نکات مثبت و منفی، و خود متن. سال از تاریخ شمسی ستون <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">created_at</span> بیرون کشیده شد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> ۱۳ تکه در حدود ۸ دقیقه خوانده شد. هیچ ردیفی به خاطر خرابی کنار نرفت و هر ۶٬۱۵۶٬۲۸۹ ردیف رسیدند.</p>
</div>

```python
parts, n_raw, t0 = [], 0, time.time()
reader = pd.read_csv(COMMENTS_CSV, usecols=[c for c in COLMAP.values() if c], dtype=str, encoding='utf-8-sig',
                     chunksize=CHUNK, on_bad_lines='skip')
for i, ch in enumerate(reader):
    n_raw += len(ch)
    ch = ch.rename(columns={c: n for n, c in COLMAP.items() if c})
    for n in missing:
        ch[n] = np.nan
    adv = norm_series(parse_col(ch['advantages']))
    dis = norm_series(parse_col(ch['disadvantages']))
    text = build_text(norm_series(ch['title']), norm_series(ch['body']), adv, dis)
    parts.append(pd.DataFrame({
        'id': ch['id'].values,
        'product_id': ch['product_id'].fillna('').values,
        'label': ch['recommendation_status'].fillna('').str.strip().str.lower().values,
        'rate': pd.to_numeric(ch['rate'], errors='coerce').astype('float32').values,
        'year': pd.to_numeric(ch['created_at'].str.extract(r'(?<!\d)(20\d\d|1[34]\d\d)(?!\d)', expand=False),
                              errors='coerce').astype('float32').values,
        'is_buyer': ch['is_buyer'].fillna('').str.strip().str.lower().values,
        'has_adv': (adv.str.len() > 0).values,
        'has_dis': (dis.str.len() > 0).values,
        'text': text,
    }))
    print(f'chunk {i + 1}: {n_raw:,} rows | {time.time() - t0:.0f}s')

df = pd.concat(parts, ignore_index=True)
del parts, reader
gc.collect()
df['is_buyer'] = df['is_buyer'].astype('category')
print(f'total rows read: {len(df):,}')
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> اگر یک متن هم در train باشد و هم در تست، مدل جواب را از قبل حفظ کرده است و عدد تست بیش از حد خوب درمی‌آید.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> ۲۵٫۵ درصد نظرها کپی یک نظر دیگرند و «عالی» به‌تنهایی ۷۲٬۹۰۳ بار آمده است. برچسب نسخه‌های یک متن هم همیشه یکی نیست و فقط در ۹۳٫۴ درصد موارد با هم توافق دارند. «معمولی» ۹٬۷۹۸ بار آمده که ۶۰ درصدش <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">no_idea</span> و ۳۶٫۷ درصدش <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">recommended</span> است. یعنی خود برچسب‌ها نویز دارند و هیچ مدلی نمی‌تواند همه‌ی نظرها را درست بگوید.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> برای مقایسه‌ی متن‌ها کلیدی ساخته شد که فقط حرف‌ها و عددها را نگه می‌دارد. از هر متن یک نسخه ماند، با برچسبی که در نسخه‌هایش اکثریت داشت. ۸٬۶۶۱ متن که هیچ برچسبی در آن‌ها اکثریت نداشت حذف شدند و ۳٬۹۱۰٬۹۲۶ متن یکتا باقی ماند.</p>

```python
t0 = time.time()
key = df['text'].str.replace(r'[\W_]+', '', regex=True)
df['key'] = key.where(key.str.len() > 0, df['text'].str.replace(r'\s+', '', regex=True))
del key

cnt = df.groupby(['key', 'y'], sort=False).size().unstack(fill_value=0)
cnt = cnt.reindex(columns=[0, 1, 2], fill_value=0)
tot = cnt.sum(axis=1)
top = cnt.max(axis=1)
info = pd.DataFrame({'n_copies': tot.astype('int32'), 'agree': (top / tot).astype('float32'),
                     'y_major': cnt.values.argmax(1).astype('int8')}, index=cnt.index)

n_rows, n_keys = len(df), len(info)
dup = info['n_copies'] > 1
print(f'rows: {n_rows:,} | unique texts: {n_keys:,} | extra copies: {n_rows - n_keys:,} ({(n_rows - n_keys) / n_rows * 100:.1f}%)')
print(f'rows whose text appears more than once: {info.loc[dup, "n_copies"].sum() / n_rows * 100:.1f}%')
print(f'label agreement among copies of the same text: {top[dup].sum() / tot[dup].sum() * 100:.1f}%')

top_txt = info.sort_values('n_copies', ascending=False).head(15).copy()
shares = (cnt.loc[top_txt.index].div(tot.loc[top_txt.index], axis=0) * 100).round(1)
shares.columns = LABELS
display(pd.concat([top_txt[['n_copies']], shares], axis=1))

uni = df.drop_duplicates('key').set_index('key').drop(columns=['y']).join(info)
ties = uni['agree'] <= 0.5
print(f'texts dropped because no label has a majority: {ties.sum():,}')
uni = uni[~ties].copy()
uni['y'] = uni.pop('y_major')
uni['label'] = np.array(LABELS)[uni['y'].values]
uni = uni.reset_index()
del df, cnt, tot, top, info
gc.collect()
print(f'unique labelled texts: {len(uni):,} | {time.time() - t0:.0f}s')
display((uni['label'].value_counts(normalize=True) * 100).round(2).reindex(LABELS))
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> نظرهای یک محصول شبیه هم‌اند. اگر بعضی در train و بعضی در تست باشند، مدل به جای «این جمله منفی است» یاد می‌گیرد «این محصول بد است».</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> هر محصول با md5 شناسه‌اش به یکی از ۱۰ سطل رفت: ۸ سطل train، یک سطل val و یک سطل تست. همه‌ی نظرهای یک محصول در یک بخش می‌مانند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> هیچ محصول و هیچ متنی بین سه بخش مشترک نیست. train با ۳٬۱۲۰٬۵۴۶ نظر، val با ۳۹۵٬۴۲۶ و تست با ۳۹۴٬۹۵۴ نظر ساخته شد و درصد کلاس‌ها در هر سه تقریباً یکی است. همه‌ی متن‌های یکتا در <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">q3_all_unique.parquet</span> ذخیره شدند.</p>

```python
pid = uni['product_id'].astype(str)
pid = pid.where(pid.str.len() > 0, 'noprod_' + uni['id'].astype(str))
prods = pd.Series(pid.unique())
bucket = dict(zip(prods, prods.map(lambda p: int(hashlib.md5(p.encode()).hexdigest(), 16) % 10)))
b = pid.map(bucket).values
uni['split'] = np.where(b < 8, 'train', np.where(b == 8, 'val', 'test'))

sets = {s: set(pid[uni['split'] == s]) for s in ('train', 'val', 'test')}
print('shared products  train-val:', len(sets['train'] & sets['val']),
      '| train-test:', len(sets['train'] & sets['test']), '| val-test:', len(sets['val'] & sets['test']))
print('every text unique across all splits:', uni['key'].is_unique)
tab = pd.crosstab(uni['split'], uni['label'], margins=True).reindex(columns=LABELS + ['All'])
display(tab)
display((pd.crosstab(uni['split'], uni['label'], normalize='index') * 100).round(2).reindex(columns=LABELS))
del sets, pid, prods, bucket, b
gc.collect()

t0 = time.time()
uni.drop(columns=['key']).to_parquet(f'{BASE}/q3_all_unique.parquet', index=False)
print(f'saved all unique labelled texts -> q3_all_unique.parquet | {time.time() - t0:.0f}s')
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> آموزش روی سه میلیون نظر در کولب عملی نیست و دو مدل به دو جور داده‌ی آموزشی نیاز دارند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> برای TF-IDF یک نمونه‌ی تصادفی ۳۰۰ هزارتایی با همان توزیع واقعی برداشته شد. برای ترنسفورمر از هر کلاس ۱۰۰ هزار نظر، یعنی ۳۰۰ هزار نظر متوازن. تصمیم‌ها روی ۴۰ هزار نظر از val گرفته می‌شوند و تست کامل، یعنی هر ۳۹۴٬۹۵۴ نظر، فقط یک بار در آخر استفاده می‌شود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> چون تست بزرگ است، بازه‌ی اطمینان Macro F1 روی آن فقط حدود ۰٫۰۰۴ پهنا دارد.</p>

```python
COLS = ['id', 'product_id', 'text', 'y', 'n_words', 'n_copies', 'agree', 'rate', 'year', 'has_adv', 'has_dis']
tr = uni[uni['split'] == 'train']
va = uni[uni['split'] == 'val']
te = uni[uni['split'] == 'test']

train_tfidf = tr.sample(n=min(N_TFIDF, len(tr)), random_state=SEED)[COLS]
train_bert = pd.concat([g.sample(n=min(PER_CLASS_BERT, len(g)), random_state=SEED) for _, g in tr.groupby('y')])
train_bert = train_bert.sample(frac=1, random_state=SEED)[COLS]
val_df = va.sample(n=min(N_VAL, len(va)), random_state=SEED)[COLS]
test_df = te.sample(frac=1, random_state=SEED)[COLS]

for name, d in [('train_tfidf', train_tfidf), ('train_bert', train_bert), ('val', val_df), ('test', test_df)]:
    d.reset_index(drop=True).to_parquet(f'{BASE}/q3_{name}.parquet', index=False)
    share = (d['y'].value_counts(normalize=True).sort_index() * 100).round(2).tolist()
    print(f'{name:12s} {len(d):>9,} rows | % per class {share}')
del tr, va, te, uni
_ = gc.collect()
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۱۷. تابع‌های ارزیابی</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> Macro F1 از روی ماتریس درهم‌ریختگی حساب می‌شود. <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">tune_bias</span> روی val دو عدد بایاس برای دو کلاس کوچک پیدا می‌کند که Macro F1 را بیشینه کند. <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">boot_ci</span> بازه‌ی اطمینان ۹۵ درصد را با ۱۰۰۰ بار نمونه‌گیری bootstrap می‌دهد و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">boot_diff</span> اختلاف دو مدل را روی همان نمونه‌ها، به صورت جفت‌شده، می‌سنجد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> مدل‌ها یا با وزن متوازن کلاس‌ها آموزش می‌بینند یا روی داده‌ی متوازن، پس کلاس‌های کوچک را بیشتر از حد پیش‌بینی می‌کنند. تنظیم بایاس روی val این را جبران می‌کند، بدون اینکه دستی به تست بخورد.</p>
</div>

```python
def conf(y, p, k=3):
    return np.bincount(np.asarray(y, dtype=np.int64) * k + np.asarray(p, dtype=np.int64), minlength=k * k).reshape(k, k)


def f1_per_class(y, p):
    cm = conf(y, p)
    tp = np.diag(cm)
    den = 2 * tp + (cm.sum(0) - tp) + (cm.sum(1) - tp)
    return np.where(den > 0, 2 * tp / np.maximum(den, 1), 0.0)


def macro_f1(y, p):
    return float(f1_per_class(y, p).mean())


def tune_bias(logp, y, grid=np.round(np.arange(-3, 3.01, 0.1), 2)):
    best_f, best_b = macro_f1(y, logp.argmax(1)), np.zeros(3)
    for b1 in grid:
        for b2 in grid:
            b = np.array([0.0, b1, b2])
            f = macro_f1(y, (logp + b).argmax(1))
            if f > best_f:
                best_f, best_b = f, b
    return best_b, best_f


def boot_ci(y, p, n=1000, seed=0):
    rng = np.random.default_rng(seed)
    y, p = np.asarray(y), np.asarray(p)
    s = [macro_f1(y[i], p[i]) for i in (rng.integers(0, len(y), len(y)) for _ in range(n))]
    return np.percentile(s, [2.5, 97.5])


def boot_diff(y, p1, p2, n=1000, seed=0):
    rng = np.random.default_rng(seed)
    y, p1, p2 = map(np.asarray, (y, p1, p2))
    d = []
    for _ in range(n):
        i = rng.integers(0, len(y), len(y))
        d.append(macro_f1(y[i], p1[i]) - macro_f1(y[i], p2[i]))
    return float(np.mean(d)), np.percentile(d, [2.5, 97.5])


def summarize(name, y, p):
    f = f1_per_class(y, p)
    lo, hi = boot_ci(y, p)
    return {'model': name, 'macro_f1': round(float(f.mean()), 4), 'ci95': f'{lo:.4f} - {hi:.4f}',
            'accuracy': round(float((np.asarray(y) == np.asarray(p)).mean()), 4),
            **{f'f1_{l}': round(float(v), 4) for l, v in zip(LABELS, f)}}


RESULTS, PRED = [], {}
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> مرجع اول همیشه کلاس پرتعداد را می‌گوید. مرجع دوم یک رگرسیون لجستیک است که فقط ستاره‌ی نظر را می‌بیند و متن را اصلاً نمی‌خواند؛ بایاس آن هم روی val تنظیم شد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> کلاس پرتعداد روی تست Macro F1 برابر ۰٫۲۸۷۵ گرفت و مدل ستاره ۰٫۷۲۹۹ (بازه‌ی ۰٫۷۲۸۱ تا ۰٫۷۳۱۷). مدل ستاره جزو جواب نیست و فقط نشان می‌دهد یک مدل متنی باید به کجا برسد تا بگوییم متن به اندازه‌ی ستاره‌ی خود خریدار اطلاعات دارد.</p>
</div>

```python
from sklearn.linear_model import LogisticRegression

maj = int(train_tfidf['y'].value_counts().idxmax())
PRED['majority'] = np.full(len(y_te), maj)
RESULTS.append(summarize('baseline: always majority class', y_te, PRED['majority']))


def rate_onehot(d):
    r = d['rate'].fillna(-1).clip(-1, 5).round().astype(int).values
    return np.eye(7)[r + 1]


rate_clf = LogisticRegression(max_iter=1000, class_weight='balanced').fit(rate_onehot(train_tfidf), train_tfidf['y'])
b_rate, _ = tune_bias(rate_clf.predict_log_proba(rate_onehot(val_df)), y_va)
p_rate = (rate_clf.predict_log_proba(rate_onehot(test_df)) + b_rate).argmax(1)
REF_RATE = summarize('reference, NOT text: star rating only', y_te, p_rate)
display(pd.DataFrame(RESULTS + [REF_RATE]))
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۲۰. چارچوب آزمایش‌های TF-IDF</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> TF-IDF چند انتخاب دارد: n-gram کلمه‌ای یا حرفی، مقدار C و پیش‌پردازش اضافه. این‌ها باید بدون نگاه به تست انتخاب شوند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">run_config</span> برای هر تنظیم، بردارساز را فقط روی train می‌سازد، رگرسیون لجستیک را برای چند C آموزش می‌دهد و Macro F1 را روی val، یک بار خام و یک بار بعد از تنظیم بایاس، گزارش می‌کند. نتیجه‌ی هر تنظیم در <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">LADDER</span> ثبت می‌شود. توکن‌ها کلمه‌ها با نیم‌فاصله‌شان و علامت‌ها هستند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">no_fields</span> و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">unify_spelling</span> دو پیش‌پردازشی هستند که در قدم‌های ۲۳ و ۲۴ امتحان شدند.</p>
</div>

```python
import warnings
import scipy.sparse as sp
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.exceptions import ConvergenceWarning
warnings.filterwarnings('ignore', category=ConvergenceWarning)

TOK = r'(?u)[\w\u200c]+|[^\w\s\u200c]'


def make_vecs(char):
    vecs = [TfidfVectorizer(token_pattern=TOK, ngram_range=(1, 2), min_df=3, max_features=300_000,
                            sublinear_tf=True, dtype=np.float32)]
    if char:
        vecs.append(TfidfVectorizer(analyzer='char_wb', ngram_range=(2, 5), min_df=5, max_features=300_000,
                                    sublinear_tf=True, dtype=np.float32))
    return vecs


def no_fields(s):
    return s.str.replace(r' \| \u0646\u06a9\u0627\u062a (?:\u0645\u062b\u0628\u062a|\u0645\u0646\u0641\u06cc): .*$', '', regex=True)


def unify_spelling(s):
    s = s.str.replace('\u200c', '', regex=False)
    return s.str.replace(r'(?<!\S)(\u0646?\u0645\u06cc) (?=\S)', r'\1', regex=True)


def run_config(name, char, prep, Cs):
    t0 = time.time()
    vecs = make_vecs(char)
    X_tr = sp.hstack([v.fit_transform(prep(train_tfidf['text'])) for v in vecs]).tocsr()
    X_va = sp.hstack([v.transform(prep(val_df['text'])) for v in vecs]).tocsr()
    print(f'{name}: {X_tr.shape[1]:,} features | {time.time() - t0:.0f}s')
    best = None
    for c in Cs:
        t0 = time.time()
        m = LogisticRegression(C=c, max_iter=300, class_weight='balanced').fit(X_tr, train_tfidf['y'].values)
        lp = m.predict_log_proba(X_va)
        bias, f_tuned = tune_bias(lp, y_va)
        f_raw = macro_f1(y_va, lp.argmax(1))
        print(f'   C={c}: val macro F1 raw {f_raw:.4f} | tuned {f_tuned:.4f} | {time.time() - t0:.0f}s')
        if best is None or f_tuned > best['f']:
            best = {'name': name, 'C': c, 'f': f_tuned, 'f_raw': f_raw, 'vecs': vecs, 'prep': prep,
                    'model': m, 'bias': bias, 'lp_va': lp, 'n_features': X_tr.shape[1]}
    LADDER.append({'config': name, 'features': best['n_features'], 'best_C': best['C'],
                   'val_macro_f1_raw': round(best['f_raw'], 4), 'val_macro_f1_tuned': round(best['f'], 4)})
    del X_tr, X_va
    gc.collect()
    return best


identity = lambda s: s
LADDER, CANDS = [], {}
print('ready')
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۲۱. n-gram کلمه‌ای، تک‌کلمه و دوکلمه</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> با ۱۹۴٬۰۰۲ ویژگی، Macro F1 روی val با بزرگ شدن C بالا رفت و در C برابر ۰٫۲۵ به ۰٫۷۱۵۱ رسید. Cهای بزرگ‌تر از ۰٫۲۵ را در اجرای قبلی امتحان کرده بودم و بهتر نشدند؛ این اجرا سمت کوچک‌تر را کامل کرد. تنظیم بایاس به‌تنهایی حدود ۰٫۰۲ اضافه کرد (از ۰٫۶۹۵۰ به ۰٫۷۱۵۱).</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> این تنظیم پایه است و بقیه با آن مقایسه می‌شوند.</p>
</div>

```python
CANDS['words'] = run_config('words 1-2 gram', char=False, prep=identity, Cs=[0.0625, 0.125, 0.25])
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۲۲. اضافه کردن n-gram حرفی</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> n-gram حرفی می‌تواند غلط‌های املایی و شکل‌های مختلف یک کلمه را بگیرد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> با ۲۹۸٬۴۵۱ ویژگی بهترین نتیجه ۰٫۷۱۳۳ شد، کمی پایین‌تر از n-gram کلمه‌ای تنها، و زمان آموزش هم تقریباً سه برابر شد. پس کنار گذاشته شد.</p>
</div>

```python
CANDS['words+chars'] = run_config('words + chars 2-5', char=True, prep=identity, Cs=[0.125, 0.25, 0.5])
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۲۳. یکسان کردن املای «می»</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> «می‌خواهم»، «میخواهم» و «می خواهم» یک کلمه‌اند ولی برای TF-IDF سه ویژگی جدا هستند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> نیم‌فاصله‌ها برداشته شدند و «می» و «نمی» به کلمه‌ی بعد چسبیدند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> تعداد ویژگی‌ها کمی کم شد (۱۹۳٬۱۳۰)، ولی Macro F1 تنظیم‌شده دقیقاً همان ۰٫۷۱۵۱ ماند. بهبودی نبود، پس تنظیم ساده‌تر ماند.</p>
</div>

```python
C_w = CANDS['words']['C']
CANDS['words+spelling'] = run_config('words + spelling unification', char=False, prep=unify_spelling, Cs=[C_w])
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۲۴. نکات مثبت و منفی واقعاً کمک می‌کنند؟</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> باید دید اضافه کردن این دو فیلد به متن ارزشش را دارد یا نه.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> همان مدل یک بار بدون این دو فیلد آموزش داده شد و دو مدل روی نظرهایی که این فیلدها را دارند مقایسه شدند. چون سلول n-gram کلمه‌ای دو بار اجرا شده بود، ردیف تکراری از جدول حذف شد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> روی کل val حذف فیلدها فقط ۰٫۰۰۵ کم کرد (۰٫۷۱۰۰ در برابر ۰٫۷۱۵۱). ولی این فیلدها فقط در ۱۷٫۱ درصد نظرها هستند (۶٬۸۳۲ نظر) و روی همین نظرها Macro F1 از ۰٫۷۳۱۷ به ۰٫۷۰۴۳ افتاد. پس فیلدها در متن ورودی ماندند.</p>
</div>

```python
abl = run_config('words, WITHOUT advantages/disadvantages', char=False, prep=no_fields, Cs=[C_w])

with_f = (val_df['has_adv'] | val_df['has_dis']).values
full = CANDS['words']
p_full = (full['lp_va'] + full['bias']).argmax(1)
p_abl = (abl['lp_va'] + abl['bias']).argmax(1)
print(f'\nval comments that have advantages or disadvantages: {with_f.sum():,} ({with_f.mean() * 100:.1f}%)')
print(f'macro F1 on them  with fields: {macro_f1(y_va[with_f], p_full[with_f]):.4f} | without: {macro_f1(y_va[with_f], p_abl[with_f]):.4f}')
print(f'macro F1 on the rest (no fields): {macro_f1(y_va[~with_f], p_full[~with_f]):.4f}')
del abl
gc.collect()
LADDER = pd.DataFrame(LADDER).drop_duplicates('config', keep='last').to_dict('records')
display(pd.DataFrame(LADDER))
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۲۵. انتخاب نهایی و سنجش روی تست</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> طبق قانون ۰٫۰۰۵، n-gram کلمه‌ای با C برابر ۰٫۲۵ انتخاب شد و فقط همین مدل یک بار روی تست سنجیده شد. بایاس‌هایی که روی val پیدا شده بود (۰، منفی ۰٫۷ و منفی ۰٫۶) روی تست اعمال شد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> Macro F1 تست بدون بایاس ۰٫۶۹۳۷ و با بایاس <b>۰٫۷۱۰۲</b> شد (بازه‌ی ۰٫۷۰۸۴ تا ۰٫۷۱۲۱). عدد val (۰٫۷۱۵۱) و تست به هم نزدیک‌اند، یعنی انتخاب‌ها روی val بیش‌برازش نکرده‌اند. ضعیف‌ترین کلاس <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">no_idea</span> است با F1 برابر ۰٫۴۷. مدل متنی ساده هنوز از مدل ستاره (۰٫۷۲۹۹) عقب است و این، هدف نوت‌بوک ۲ را مشخص می‌کند.</p>
</div>

```python
best_key = max(CANDS, key=lambda k: CANDS[k]['f'])
if CANDS[best_key]['f'] - CANDS['words']['f'] < 0.005:
    best_key = 'words'
TF = CANDS[best_key]
print('chosen on validation:', TF['name'], '| C =', TF['C'])

t0 = time.time()
X_te = sp.hstack([v.transform(TF['prep'](test_df['text'])) for v in TF['vecs']]).tocsr()
lp_te_tfidf = TF['model'].predict_log_proba(X_te)
lp_va_tfidf, bias_tfidf = TF['lp_va'], TF['bias']
print(f'test transformed and predicted in {time.time() - t0:.0f}s | bias {bias_tfidf}')

PRED['tfidf'] = (lp_te_tfidf + bias_tfidf).argmax(1)
RESULTS.append(summarize(f'TF-IDF + LR, {TF["name"]} (raw)', y_te, lp_te_tfidf.argmax(1)))
RESULTS.append(summarize(f'TF-IDF + LR, {TF["name"]} (bias tuned on val)', y_te, PRED['tfidf']))
display(pd.DataFrame(RESULTS + [REF_RATE]))

np.save(f'{OUT}/logp_val_tfidf.npy', lp_va_tfidf)
np.save(f'{OUT}/logp_test_tfidf.npy', lp_te_tfidf)
pd.DataFrame(LADDER).to_csv(f'{OUT}/tfidf_ladder.csv', index=False)
del X_te
_ = gc.collect()
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۲۶. کلمه‌هایی که مدل برای هر کلاس مهم دانست</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> ۲۰ ویژگی با بیشترین ضریب برای هر کلاس استخراج شد. ویژگی‌هایی که فقط جداکننده بودند از نمایش کنار رفتند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> برای <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">recommended</span> کلمه‌هایی مثل «عالی»، «خوبه» و «راضیم» آمده‌اند و برای <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">not_recommended</span> «افتضاح»، «نخرید»، «تقلبی»، «مرجوع» و «فیک». فهرست <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">no_idea</span> جالب‌تر است: «معمولی»، «ولی»، «اما»، «نمیدونم»، «مطمئن نیستم»، «فعلا» و «نکردم». یعنی این کلاس بیشتر نظرهای دوطرفه یا زودهنگام است، جمله‌هایی که یک «ولی» وسطشان دارند. TF-IDF ترتیب کلمه‌ها را نمی‌بیند و برای همین در این کلاس ضعیف است.</p>
