<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> همان تابع‌های نوت‌بوک ۱ است تا عددهای دو نوت‌بوک دقیقاً با هم قابل مقایسه باشند.</p>

```python
def conf(y, p, k=3):
    return np.bincount(np.asarray(y, dtype=np.int64) * k + np.asarray(p, dtype=np.int64), minlength=k * k).reshape(k, k)


def f1_per_class(y, p):
    cm = conf(y, p)
    tp = np.diag(cm)
    den = 2 * tp + (cm.sum(0) - tp) + (cm.sum(1) - tp)
    return np.where(den > 0, 2 * tp / np.maximum(den, 1), 0.0)


def macro_f1(y, p):
    return float(f1_per_class(y, p).mean())


def tune_bias(logp, y, grid=np.round(np.arange(-3, 3.01, 0.1), 2)):
    best_f, best_b = macro_f1(y, logp.argmax(1)), np.zeros(3)
    for b1 in grid:
        for b2 in grid:
            b = np.array([0.0, b1, b2])
            f = macro_f1(y, (logp + b).argmax(1))
            if f > best_f:
                best_f, best_b = f, b
    return best_b, best_f


def boot_ci(y, p, n=1000, seed=0):
    rng = np.random.default_rng(seed)
    y, p = np.asarray(y), np.asarray(p)
    s = [macro_f1(y[i], p[i]) for i in (rng.integers(0, len(y), len(y)) for _ in range(n))]
    return np.percentile(s, [2.5, 97.5])


def boot_diff(y, p1, p2, n=1000, seed=0):
    rng = np.random.default_rng(seed)
    y, p1, p2 = map(np.asarray, (y, p1, p2))
    d = []
    for _ in range(n):
        i = rng.integers(0, len(y), len(y))
        d.append(macro_f1(y[i], p1[i]) - macro_f1(y[i], p2[i]))
    return float(np.mean(d)), np.percentile(d, [2.5, 97.5])


def summarize(name, y, p):
    f = f1_per_class(y, p)
    lo, hi = boot_ci(y, p)
    return {'model': name, 'macro_f1': round(float(f.mean()), 4), 'ci95': f'{lo:.4f} - {hi:.4f}',
            'accuracy': round(float((np.asarray(y) == np.asarray(p)).mean()), 4),
            **{f'f1_{l}': round(float(v), 4) for l, v in zip(LABELS, f)}}
```

<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۵. ابزار آموزش</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> برای فاین‌تیون می‌خواستم همه‌ی جزئیات آموزش، از زمان‌بندی نرخ یادگیری تا ذخیره‌ی بهترین ایپاک، دیده و کنترل شود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> به جای <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">Trainer</span> یک حلقه‌ی آموزش ساده نوشته شد: AdamW با weight decay برابر ۰٫۰۱ (به جز بایاس‌ها و لایه‌های نرمال‌سازی)، warmup شش درصد و بعد کاهش خطی نرخ یادگیری، fp16 با <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">GradScaler</span> و بریدن گرادیان در ۱. بچ‌ها بر اساس طول گروه‌بندی می‌شوند تا padding کم شود. بعد از هر ایپاک loss و Macro F1 روی val حساب می‌شود و اگر بهتر بود، مدل ذخیره می‌شود.</p>
</div>

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

USE_AMP = device.type == 'cuda'


def encode(tok, texts, max_len):
    return tok(list(texts), truncation=True, max_length=max_len)['input_ids']


def collate(seqs, pad_id):
    m = max(len(s) for s in seqs)
    ids = torch.full((len(seqs), m), pad_id, dtype=torch.long)
    att = torch.zeros((len(seqs), m), dtype=torch.long)
    for i, s in enumerate(seqs):
        ids[i, :len(s)] = torch.tensor(s)
        att[i, :len(s)] = 1
    return ids.to(device), att.to(device)


def length_batches(lengths, bs, rng):
    idx = rng.permutation(len(lengths))
    out = []
    for s in range(0, len(idx), bs * 50):
        c = idx[s:s + bs * 50]
        c = c[np.argsort(lengths[c], kind='stable')]
        out += [c[i:i + bs] for i in range(0, len(c), bs)]
    return [out[i] for i in rng.permutation(len(out))]


@torch.no_grad()
def predict_logp(model, tok, texts, max_len, bs=256):
    model.eval()
    seqs = encode(tok, texts, max_len)
    pad = tok.pad_token_id if tok.pad_token_id is not None else 0
    order = np.argsort([-len(s) for s in seqs], kind='stable')
    res = np.zeros((len(seqs), 3), dtype=np.float32)
    for s in range(0, len(order), bs):
        idx = order[s:s + bs]
        ids, att = collate([seqs[i] for i in idx], pad)
        with torch.autocast(device_type=device.type, dtype=torch.float16, enabled=USE_AMP):
            logits = model(input_ids=ids, attention_mask=att).logits
        res[idx] = torch.log_softmax(logits.float(), -1).cpu().numpy()
    return res


def train_model(model_name, texts, labels, max_len, epochs, lr, bs=32, warmup=0.06, wd=0.01,
                eval_texts=None, eval_y=None, save_dir=None, log_every=500, seed=SEED):
    torch.manual_seed(seed)
    tok = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSequenceClassification.from_pretrained(
        model_name, num_labels=3, id2label=dict(enumerate(LABELS)), label2id=L2I).to(device)
    pad = tok.pad_token_id if tok.pad_token_id is not None else 0
    t0 = time.time()
    seqs = encode(tok, texts, max_len)
    lens = np.array([len(s) for s in seqs])
    y = np.asarray(labels)
    n_params = sum(p.numel() for p in model.parameters()) / 1e6
    print(f'{model_name}: {n_params:.1f}M params | {len(seqs):,} texts, mean {lens.mean():.1f} tokens | {time.time() - t0:.0f}s')

    decay = [p for n, p in model.named_parameters() if not ('bias' in n or 'norm' in n.lower())]
    no_decay = [p for n, p in model.named_parameters() if 'bias' in n or 'norm' in n.lower()]
    opt = torch.optim.AdamW([{'params': decay, 'weight_decay': wd},
                             {'params': no_decay, 'weight_decay': 0.0}], lr=lr)
    total = math.ceil(len(seqs) / bs) * epochs
    warm = max(1, int(warmup * total))
    sched = torch.optim.lr_scheduler.LambdaLR(
        opt, lambda s: min((s + 1) / warm, max(0.0, (total - s) / max(1, total - warm))))
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP)

    rng = np.random.default_rng(seed)
    history, best, step, t0 = [], -1.0, 0, time.time()
    for ep in range(epochs):
        model.train()
        run, seen = 0.0, 0
        for b in length_batches(lens, bs, rng):
            ids, att = collate([seqs[i] for i in b], pad)
            yb = torch.tensor(y[b], dtype=torch.long, device=device)
            with torch.autocast(device_type=device.type, dtype=torch.float16, enabled=USE_AMP):
                logits = model(input_ids=ids, attention_mask=att).logits
            loss = torch.nn.functional.cross_entropy(logits.float(), yb)
            opt.zero_grad(set_to_none=True)
            scaler.scale(loss).backward()
            scaler.unscale_(opt)
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(opt)
            scaler.update()
            sched.step()
            run += loss.item() * len(b)
            seen += len(b)
            step += 1
            if step % log_every == 0:
                print(f'  epoch {ep + 1} step {step:,}/{total:,} | loss {run / seen:.4f} | '
                      f'lr {sched.get_last_lr()[0]:.2e} | {(time.time() - t0) / 60:.1f} min')
        rec = {'epoch': ep + 1, 'train_loss': round(run / seen, 4), 'minutes': round((time.time() - t0) / 60, 1)}
        if eval_texts is not None:
            lp = predict_logp(model, tok, eval_texts, max_len)
            rec['val_loss'] = round(float(-lp[np.arange(len(eval_y)), eval_y].mean()), 4)
            rec['val_f1_raw'] = round(macro_f1(eval_y, lp.argmax(1)), 4)
            rec['val_f1_tuned'] = round(tune_bias(lp, eval_y)[1], 4)
            if save_dir and rec['val_f1_tuned'] > best:
                best = rec['val_f1_tuned']
                model.save_pretrained(save_dir)
                tok.save_pretrained(save_dir)
                rec['saved'] = True
        history.append(rec)
        print(' ', rec)
    return model, tok, history


print('ready')
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> روی ۵۰۰۰ نظر، FaBERT کمترین تعداد توکن به ازای هر کلمه را داشت (۱٫۳۳). سهم توکن ناشناخته برای TookaBERT و FaBERT برابر ۰٫۰۳۸ درصد بود و برای ParsBERT برابر ۰٫۱۴۹ درصد. صدک ۹۹ طول نظرها برای هر سه مدل ۱۰۵ توکن یا کمتر بود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> حداکثر طول ورودی ۱۲۸ توکن گذاشته شد. این طول ۹۹ درصد نظرها را کامل جا می‌دهد و آموزش را از طول ۲۵۶ یا ۵۱۲ خیلی سریع‌تر می‌کند.</p>
</div>

```python
CANDIDATES = ['PartAI/TookaBERT-Base', 'sbunlp/fabert', 'HooshvareLab/bert-fa-zwnj-base']
smp = train_bert['text'].sample(5000, random_state=SEED).tolist()
n_words = np.array([len(t.split()) for t in smp])
TOK_ROWS = []
for name in CANDIDATES:
    try:
        tok = AutoTokenizer.from_pretrained(name)
        ids = tok(smp, truncation=False)['input_ids']
        lens = np.array([len(x) for x in ids])
        unk = tok.unk_token_id
        unk_rate = np.mean([t == unk for x in ids for t in x]) if unk is not None else 0.0
        TOK_ROWS.append({'model': name, 'vocab': tok.vocab_size, 'tokens_per_word': round(float((lens / n_words).mean()), 2),
                         'unk_pct': round(100 * float(unk_rate), 3), 'p50': int(np.percentile(lens, 50)),
                         'p95': int(np.percentile(lens, 95)), 'p99': int(np.percentile(lens, 99)), 'max': int(lens.max())})
        print(name, '->', tok.tokenize(smp[0])[:20])
    except Exception as e:
        print(name, 'FAILED:', str(e)[:300])
display(pd.DataFrame(TOK_ROWS))
MAX_LEN = int(min(256, max(64, math.ceil(max(r['p99'] for r in TOK_ROWS) / 32) * 32)))
print('MAX_LEN =', MAX_LEN)
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۷. غربال سه مدل با شرایط یکسان</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> آموزش کامل هر سه مدل وقت زیادی می‌گیرد. اول باید با هزینه‌ی کم دید کدام امیدوارکننده‌تر است.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> هر مدل روی ۳۰ هزار نظر متوازن (از هر کلاس ۱۰ هزار) یک ایپاک با نرخ یادگیری <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">3e-5</span> آموزش دید و روی همان ۴۰ هزار نظر val سنجیده شد. هر آموزش حدود دو دقیقه طول کشید.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> FaBERT با Macro F1 تنظیم‌شده‌ی ۰٫۷۳۶۲ اول شد، TookaBERT با ۰٫۷۳۳۰ دوم و ParsBERT با ۰٫۷۱۵۲ سوم. FaBERT با همین آموزش کوچک از TF-IDF روی val (۰٫۷۱۵۱) جلو افتاد.</p>
</div>

```python
SCREEN_PER_CLASS = 10_000
screen = pd.concat([g.sample(n=SCREEN_PER_CLASS, random_state=SEED) for _, g in train_bert.groupby('y')])
screen = screen.sample(frac=1, random_state=SEED)

SCREEN = []
for name in [r['model'] for r in TOK_ROWS]:
    model = tok = None
    t0 = time.time()
    try:
        model, tok, hist = train_model(name, screen['text'], screen['y'].values, MAX_LEN, epochs=1, lr=3e-5,
                                       eval_texts=val_df['text'], eval_y=y_va, log_every=300)
        SCREEN.append({'model': name, **hist[-1], 'total_minutes': round((time.time() - t0) / 60, 1)})
    except Exception as e:
        print(name, 'FAILED:', repr(e)[:300])
    del model, tok
    gc.collect()
    torch.cuda.empty_cache()

SCREEN_DF = pd.DataFrame(SCREEN).sort_values('val_f1_tuned', ascending=False)
display(SCREEN_DF)
SCREEN_DF.to_csv(f'{OUT}/screening.csv', index=False)
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۸. انتخاب مدل و آموزش کامل</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> فاصله‌ی FaBERT و TookaBERT فقط ۰٫۰۰۳۲ بود، کمتر از حد ۰٫۰۰۵. با این فاصله نمی‌شود گفت کدام واقعاً بهتر است.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> برای حالت تساوی از قبل قانونی در کد نوشته شد: بین مدل‌هایی که تا ۰٫۰۰۵ با اولی فاصله دارند، مدلی انتخاب می‌شود که توکن ناشناخته‌ی کمتری دارد و بعد توکن کمتری به ازای هر کلمه. FaBERT انتخاب شد و روی ۳۰۰ هزار نظر متوازن دو ایپاک آموزش دید (۱۸٬۷۵۰ گام با batch برابر ۳۲).</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> بعد از ایپاک اول Macro F1 تنظیم‌شده روی val به ۰٫۷۵۰۸ رسید. در ایپاک دوم loss آموزش از ۰٫۵۶۶ به ۰٫۴۷۶ رسید، ولی loss روی val از ۰٫۴۴۰ به ۰٫۴۴۵ بالا رفت و Macro F1 هم کمی کم شد (۰٫۷۵۰۰). مدل داشت بیش‌برازش می‌کرد، پس نسخه‌ی ایپاک اول که خودکار ذخیره شده بود نگه داشته شد. کل آموزش ۳۳ دقیقه طول کشید.</p>
</div>

```python
s = SCREEN_DF.reset_index(drop=True)
if s.loc[0, 'val_f1_tuned'] - s.loc[1, 'val_f1_tuned'] < 0.005:
    tied = s[s['val_f1_tuned'] >= s.loc[0, 'val_f1_tuned'] - 0.005]
    tied = tied.merge(pd.DataFrame(TOK_ROWS)[['model', 'tokens_per_word', 'unk_pct']], on='model')
    BEST_NAME = tied.sort_values(['unk_pct', 'tokens_per_word', 'val_loss']).iloc[0]['model']
else:
    BEST_NAME = s.loc[0, 'model']
FT_DIR = f'{MODELS}/{BEST_NAME.split("/")[-1]}_full'
print('chosen:', BEST_NAME, '->', FT_DIR)

model, tok, HIST = train_model(BEST_NAME, train_bert['text'], train_bert['y'].values, MAX_LEN, epochs=2, lr=3e-5,
                               eval_texts=val_df['text'], eval_y=y_va, save_dir=FT_DIR, log_every=1000)
json.dump({'model_name': BEST_NAME, 'max_len': MAX_LEN, 'train_size': len(train_bert), 'epochs': 2,
           'lr': 3e-5, 'batch': 32, 'history': HIST}, open(f'{OUT}/bert_full_history.json', 'w'), indent=1)
display(pd.DataFrame(HIST))
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۹. سنجش FaBERT روی تست</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> بهترین نسخه از درایو بارگذاری شد و برای ۴۳۴٬۹۵۴ نظر val و تست در ۱۹۹ ثانیه احتمال حساب شد. بایاس‌ها روی val پیدا شدند (۰، منفی ۱٫۱ و منفی ۱٫۰) و همان‌ها روی تست اعمال شدند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> Macro F1 تست بدون بایاس ۰٫۷۲۸۱ و با بایاس <b>۰٫۷۴۷۱</b> شد (بازه‌ی ۰٫۷۴۵۳ تا ۰٫۷۴۸۹). FaBERT از TF-IDF ۰٫۰۳۶۹ و از مدل ستاره ۰٫۰۱۷۲ بهتر است و بازه‌ی اطمینان هر دو اختلاف صفر را شامل نمی‌شود. یعنی مدلی که فقط متن را می‌خواند حالا از ستاره‌ی خود خریدار هم دقیق‌تر است. <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">no_idea</span> هنوز ضعیف‌ترین کلاس است، ولی F1 آن از ۰٫۴۷۳ در TF-IDF به ۰٫۵۳۲ رسید. فاصله‌ی val (۰٫۷۵۰۸) و تست هم کم است.</p>
</div>

```python
del model
gc.collect()
torch.cuda.empty_cache()

best = AutoModelForSequenceClassification.from_pretrained(FT_DIR).to(device)
best_tok = AutoTokenizer.from_pretrained(FT_DIR)
NAME = BEST_NAME.split('/')[-1]

t0 = time.time()
lp_va_bert = predict_logp(best, best_tok, val_df['text'], MAX_LEN)
lp_te_bert = predict_logp(best, best_tok, test_df['text'], MAX_LEN)
print(f'predicted {len(val_df) + len(test_df):,} texts in {time.time() - t0:.0f}s')

bias_bert, f_va = tune_bias(lp_va_bert, y_va)
print('val macro F1 raw', round(macro_f1(y_va, lp_va_bert.argmax(1)), 4), '| tuned', round(f_va, 4), '| bias', bias_bert)

PRED['bert'] = (lp_te_bert + bias_bert).argmax(1)
RESULTS.append(summarize(f'{NAME} fine-tuned (raw)', y_te, lp_te_bert.argmax(1)))
RESULTS.append(summarize(f'{NAME} fine-tuned (bias tuned on val)', y_te, PRED['bert']))
display(pd.DataFrame(RESULTS + [REF_RATE]))

d, (lo, hi) = boot_diff(y_te, PRED['bert'], PRED['tfidf'])
print(f'{NAME} minus TF-IDF:         {d:+.4f}  (95% CI {lo:+.4f} to {hi:+.4f})')
p_rate = np.load(f'{OUT}/pred_test_rate.npy')
d2, (lo2, hi2) = boot_diff(y_te, PRED['bert'], p_rate)
print(f'{NAME} minus star reference: {d2:+.4f}  (95% CI {lo2:+.4f} to {hi2:+.4f})')

np.save(f'{OUT}/logp_val_bert.npy', lp_va_bert)
np.save(f'{OUT}/logp_test_bert.npy', lp_te_bert)
json.dump({'model_name': BEST_NAME, 'max_len': MAX_LEN, 'labels': LABELS, 'bias': bias_bert.tolist()},
          open(f'{FT_DIR}/q3_config.json', 'w'), indent=1)
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">۱۰. ترکیب با TF-IDF</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> شاید TF-IDF چیزهایی ببیند که FaBERT نمی‌بیند و ترکیب دو مدل بهتر شود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> میانگین وزن‌دار احتمال‌های لگاریتمی دو مدل برای وزن‌های ۰ تا ۱ روی val امتحان شد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> بهترین وزن FaBERT برابر ۰٫۷۵ بود، با Macro F1 برابر ۰٫۷۵۳۰ روی val در برابر ۰٫۷۵۰۸ برای FaBERT تنها. این ۰٫۰۰۲۲ بهبود از حد ۰٫۰۰۵ کمتر است و ترکیب پذیرفته نشد. روی تست هم فقط ۰٫۰۰۰۸ اضافه کرد؛ آن‌قدر کم که نگه داشتن یک مدل اضافه در عمل ارزشش را ندارد.</p>
</div>

```python
ENS = []
for w in np.round(np.arange(0, 1.01, 0.05), 2):
    b, f = tune_bias(w * lp_va_bert + (1 - w) * lp_va_tfidf, y_va)
    ENS.append({'w_bert': float(w), 'val_f1_tuned': round(f, 4), 'bias': b})
ENS_DF = pd.DataFrame(ENS)
best_e = ENS_DF.loc[ENS_DF['val_f1_tuned'].idxmax()]
W_BERT, bias_ens = float(best_e['w_bert']), best_e['bias']
f_bert_only = float(ENS_DF.loc[ENS_DF['w_bert'] == 1.0, 'val_f1_tuned'].iloc[0])
print(f'best FaBERT weight {W_BERT} | val {best_e["val_f1_tuned"]:.4f} vs FaBERT alone {f_bert_only:.4f} | bias {bias_ens}')
print('ensemble adopted:', best_e['val_f1_tuned'] - f_bert_only >= 0.005)

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(ENS_DF['w_bert'], ENS_DF['val_f1_tuned'], marker='o')
ax.set_xlabel('weight of FaBERT (rest = TF-IDF)')
ax.set_ylabel('val macro F1 (tuned)')
plt.tight_layout()
plt.savefig(f'{OUT}/ensemble_weight.png', dpi=150)
plt.show()

lp_te_ens = W_BERT * lp_te_bert + (1 - W_BERT) * lp_te_tfidf
PRED['ensemble'] = (lp_te_ens + bias_ens).argmax(1)
RESULTS.append(summarize(f'ensemble: {W_BERT:.2f} {NAME} + {1 - W_BERT:.2f} TF-IDF (tuned on val)', y_te, PRED['ensemble']))
display(pd.DataFrame(RESULTS + [REF_RATE]))
d, (lo, hi) = boot_diff(y_te, PRED['ensemble'], PRED['bert'])
print(f'ensemble minus {NAME}: {d:+.4f}  (95% CI {lo:+.4f} to {hi:+.4f})')

json.dump({'results': RESULTS, 'ref_rate': REF_RATE, 'w_bert': W_BERT, 'bias_ens': bias_ens.tolist(),
           'bias_bert': bias_bert.tolist(), 'best_model': BEST_NAME, 'history': HIST},
          open(f'{OUT}/phaseC_results.json', 'w'), indent=1)
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> دما ۱٫۰۵۵ و بایاس‌ها منفی ۱٫۸۶۶ و منفی ۱٫۸۷۸ شدند. log loss تست از ۰٫۴۴۱۶ به ۰٫۳۱۶۲ و ECE از ۰٫۰۳۵۰ به ۰٫۰۰۳۹ رسید؛ خطای کالیبراسیون حدود ۹ برابر کم شد. این بایاس‌ها با بایاس‌های قدم ۹ فرق دارند: آن‌ها برای بیشترین Macro F1 تنظیم شده‌اند و این‌ها برای احتمال درست.</p>

```python
from scipy.optimize import minimize
from scipy.special import log_softmax


def nll(params, lp, y):
    T, b1, b2 = params
    z = log_softmax(lp / T + np.array([0.0, b1, b2]), axis=1)
    return -z[np.arange(len(y)), y].mean()


def ece(prob, y, bins=15):
    conf, acc = prob.max(1), (prob.argmax(1) == y)
    edges, e, rows = np.linspace(0, 1, bins + 1), 0.0, []
    for lo, hi in zip(edges[:-1], edges[1:]):
        m = (conf > lo) & (conf <= hi)
        if m.any():
            e += m.mean() * abs(acc[m].mean() - conf[m].mean())
            rows.append((conf[m].mean(), acc[m].mean(), int(m.sum())))
    return e, rows


res = minimize(nll, x0=[1.0, 0.0, 0.0], args=(lp_va_bert, y_va), method='Nelder-Mead')
T_CAL, B_CAL = float(res.x[0]), np.array([0.0, res.x[1], res.x[2]])


def calibrate(lp):
    return np.exp(log_softmax(lp / T_CAL + B_CAL, axis=1))


p_raw = np.exp(lp_te_bert)
p_cal = calibrate(lp_te_bert)
e_raw, rows_raw = ece(p_raw, y_te)
e_cal, rows_cal = ece(p_cal, y_te)
print(f'temperature {T_CAL:.3f} | bias {B_CAL.round(3)}')
print(f'test NLL raw {nll([1, 0, 0], lp_te_bert, y_te):.4f} -> calibrated {nll(res.x, lp_te_bert, y_te):.4f}')
print(f'test ECE raw {e_raw:.4f} -> calibrated {e_cal:.4f}')

fig, ax = plt.subplots(figsize=(4.5, 4.5))
ax.plot([0, 1], [0, 1], '--', color='gray', label='perfect')
for rows, lab in [(rows_raw, 'raw'), (rows_cal, 'calibrated')]:
    ax.plot([r[0] for r in rows], [r[1] for r in rows], marker='o', label=lab)
ax.set_xlabel('model confidence')
ax.set_ylabel('actual accuracy')
ax.legend()
plt.tight_layout()
plt.savefig(f'{OUT}/calibration.png', dpi=150)
plt.show()
json.dump({'T': T_CAL, 'bias': B_CAL.tolist(), 'ece_raw': e_raw, 'ece_cal': e_cal},
          open(f'{FT_DIR}/calibration.json', 'w'), indent=1)
```

```python
P_va = calibrate(lp_va_bert)
print(f'val: {len(y_va)} nazar | FaBERT Macro F1 = {macro_f1(y_va, (lp_va_bert + bias_bert).argmax(1)):.4f}')
print(f'ECE roo val: kalibre-nashode {ece(np.exp(lp_va_bert), y_va)[0]:.4f} | kalibre-shode {ece(P_va, y_va)[0]:.4f}')

R_N, R_MAX = 2000, 0.30
rng_r = np.random.default_rng(SEED + 7)
r_idx = np.sort(rng_r.choice(len(y_va), R_N, replace=False))
conf_r = P_va[r_idx].max(1)
order = np.argsort(conf_r, kind='stable')
n_ask = int(R_N * R_MAX)
ask_idx = np.sort(r_idx[order[:n_ask]])
low = np.isin(r_idx, ask_idx)
pred_r = (lp_va_bert[r_idx] + bias_bert).argmax(1)
y_r = y_va[r_idx]
print(f'\n{R_N} nazar-e val | {n_ask} kam-etminan-tarin baraye LLM (etminan-e kalibre-shode <= {conf_r[order[n_ask - 1]]:.3f})')
print(f'deghat-e FaBERT: kam-etminan-ha {(pred_r[low] == y_r[low]).mean():.3f} | baghie {(pred_r[~low] == y_r[~low]).mean():.3f}')
print('tozi-e kelas-ha: kam-etminan-ha', {LABELS[k]: int((y_r[low] == k).sum()) for k in range(3)},
      '| hame-ye 2000', {LABELS[k]: int((y_r == k).sum()) for k in range(3)})

val_txt = val_df['text'].to_numpy()
JOB_VAL = {'created_utc': pd.Timestamp.now(tz='UTC').isoformat(), 'seed': SEED, 'split': 'val',
           'labels': LABELS, 'system': SYSTEM, 'few': FEW,
           'items': [{'id': int(i), 'text': str(val_txt[i])} for i in ask_idx]}
VAL_JOB_PATH = f'{OUT}/llm_job_val.json'
with open(VAL_JOB_PATH, 'w', encoding='utf-8') as f:
    json.dump(JOB_VAL, f, ensure_ascii=False, indent=1)
np.save(f'{OUT}/router_val_idx.npy', r_idx)
np.save(f'{OUT}/router_val_ask_idx.npy', ask_idx)
print('\nzakhire shod:', VAL_JOB_PATH, '|', len(JOB_VAL['items']), 'nazar | prompt va few-shot = hamoon-e test')
files.download(VAL_JOB_PATH)
```

```python
rows_v = []
with open(f'{OUT}/llm_cache_val.jsonl', encoding='utf-8') as f:
    for line in f:
        try:
            rows_v.append(json.loads(line))
        except json.JSONDecodeError:
            pass
cv = pd.DataFrame(rows_v)
okv = cv[cv['err'].isna()].drop_duplicates(['model', 'mode', 'id'], keep='last')
print(f'cache-e val: {len(cv)} radif | movafagh-e yekta: {len(okv)}')

MODEL = 'openai/gpt-4o-mini'
MODES = ['zero', 'few']
pos_r = {int(i): k for k, i in enumerate(r_idx)}
LLM_VAL = {}
for mode in MODES:
    g = okv[(okv['model'] == MODEL) & (okv['mode'] == mode)]
    p = np.full(len(r_idx), -1)
    for i, lab in zip(g['id'].astype(int), g['label']):
        if i in pos_r and lab in L2I:
            p[pos_r[i]] = L2I[lab]
    LLM_VAL[mode] = p
    print(f'{mode}: javab baraye {int((p[low] >= 0).sum())} az {int(low.sum())} nazar-e kam-etminan')

def route(pred, conf_order, llm_pred, k):
    h = pred.copy()
    sel = conf_order[:k]
    h[sel] = np.where(llm_pred[sel] >= 0, llm_pred[sel], h[sel])
    return h

FRACS = [0, 0.025, 0.05, 0.075, 0.10, 0.125, 0.15, 0.20, 0.25, 0.30]
base_val = macro_f1(y_r, pred_r)
grid = []
for mode in MODES:
    for fr in FRACS:
        k = int(round(fr * len(r_idx)))
        m = macro_f1(y_r, route(pred_r, order, LLM_VAL[mode], k))
        grid.append({'mode': mode, 'frac': fr, 'k': k, 'macro_f1': m, 'gain': m - base_val})
VAL_GRID = pd.DataFrame(grid)
print(f'\nval (2000 nazar), Macro F1 bar hasb-e sahm-e nazar-haye ferestade-shode be LLM (FaBERT tanha = {base_val:.4f}):')
display(VAL_GRID.pivot(index='frac', columns='mode', values='macro_f1').round(4))

best = VAL_GRID.sort_values(['macro_f1', 'frac'], ascending=[False, True]).iloc[0]
MODE_R, K_R, GAIN_R = best['mode'], int(best['k']), float(best['gain'])
ADOPT = GAIN_R >= 0.005
cs = conf_r[order]
TAU = 0.0 if K_R == 0 else float((cs[K_R - 1] + cs[K_R]) / 2)
h_val = route(pred_r, order, LLM_VAL[MODE_R], K_R)
IDXV = np.random.default_rng(SEED + 8).integers(0, len(r_idx), (2000, len(r_idx)))
dv = np.array([macro_f1(y_r[ix], h_val[ix]) - macro_f1(y_r[ix], pred_r[ix]) for ix in IDXV])
print(f'\nbehtarin roo val: halat={MODE_R} | sahm={best["frac"]:.3f} ({K_R} nazar) | astane-ye etminan={TAU:.4f} | '
      f'behbood={GAIN_R:+.4f} (bazeh {np.percentile(dv, 2.5):+.4f} ta {np.percentile(dv, 97.5):+.4f})')
print('tasmim (ghanoon-e 0.005):', 'masiryab paziroofte shod' if ADOPT else 'masiryab paziroofte nashod, FaBERT tanha mimoone')

name_r = f'llm_{MODEL.split("/")[-1]}_{MODE_R}'
conf_te600 = calibrate(lp_te_bert[llm_idx]).max(1)
route_te = conf_te600 < TAU
hyb = SYS['fabert'].copy()
hyb[route_te] = SYS[name_r][route_te]
f_h, f_b = f1_per_class(y600, hyb), f1_per_class(y600, SYS['fabert'])
d = np.array([macro_f1(y600[ix], hyb[ix]) - macro_f1(y600[ix], SYS['fabert'][ix]) for ix in IDX])
frac_full = float((calibrate(lp_te_bert).max(1) < TAU).mean())
usd_1000 = frac_full * LLM_INFO[name_r]['usd_per_1000_server']
print(f'\ntest (600 nazar, yek bar): {int(route_te.sum())} nazar be LLM | sahm roo kol-e test = {frac_full:.3f}')
print(f'Macro F1: FaBERT {f_b.mean():.4f} | masiryab {f_h.mean():.4f} | ekhtelaf {f_h.mean() - f_b.mean():+.4f} '
      f'(bazeh {np.percentile(d, 2.5):+.4f} ta {np.percentile(d, 97.5):+.4f})')
print('F1 har kelas (FaBERT -> masiryab):', {LABELS[k]: f'{f_b[k]:.3f} -> {f_h[k]:.3f}' for k in range(3)})
print(f'hazine-ye LLM baraye 1000 nazar: ${usd_1000:.4f} | darkhast-e LLM baraye 1000 nazar: {1000 * frac_full:.0f}')

order_te = np.argsort(conf_te600, kind='stable')
desc = []
for mode in MODES:
    nm = f'llm_{MODEL.split("/")[-1]}_{mode}'
    for fr in FRACS:
        k = int(round(fr * len(llm_idx)))
        desc.append({'mode': mode, 'frac': fr, 'macro_f1': macro_f1(y600, route(SYS['fabert'], order_te, SYS[nm], k))})
print('\nfaghat tosifi (baraye entekhab estefade nashode): hamin monhani roo 600 nazar-e test')
display(pd.DataFrame(desc).pivot(index='frac', columns='mode', values='macro_f1').round(4))

with open(f'{OUT}/router_eval.json', 'w') as f:
    json.dump({'calibration': {'T': T_CAL, 'bias': B_CAL.tolist(), 'form': 'softmax(lp / T + bias)'},
               'val': {'n': int(len(r_idx)), 'n_llm': int(low.sum()), 'fabert_macro_f1': float(base_val),
                       'grid': VAL_GRID.to_dict('records')},
               'chosen': {'mode': MODE_R, 'frac_val': float(best['frac']), 'tau': TAU, 'gain_val': GAIN_R,
                          'gain_val_ci': [float(np.percentile(dv, 2.5)), float(np.percentile(dv, 97.5))],
                          'adopted': bool(ADOPT)},
               'test600': {'n_routed': int(route_te.sum()), 'fabert': float(f_b.mean()), 'router': float(f_h.mean()),
                           'diff_ci': [float(np.percentile(d, 2.5)), float(np.percentile(d, 97.5))],
                           'frac_full_test': frac_full, 'usd_per_1000': usd_1000}},
              f, indent=1, default=float)
print('\nzakhire shod: router_eval.json')
```

```python
cfg = json.load(open(f'{FT_DIR}/q3_config.json'))
MAX_LEN, BIAS_F1 = int(cfg['max_len']), np.asarray(cfg['bias'], dtype=np.float32)
tok = AutoTokenizer.from_pretrained(FT_DIR)

def logp_batch(model, texts, bs, amp):
    dev = next(model.parameters()).device
    ids = tok(list(texts), truncation=True, max_length=MAX_LEN)['input_ids']
    order_ = np.argsort([len(x) for x in ids], kind='stable')
    out = np.zeros((len(ids), 3), dtype=np.float32)
    for s in range(0, len(order_), bs):
        sel = order_[s:s + bs]
        b = tok.pad({'input_ids': [ids[i] for i in sel]}, return_tensors='pt')
        with torch.inference_mode(), torch.autocast(dev.type, dtype=torch.float16, enabled=amp):
            lo = model(input_ids=b['input_ids'].to(dev), attention_mask=b['attention_mask'].to(dev)).logits
        out[sel] = torch.log_softmax(lo.float(), -1).cpu().numpy()
    return out

def label_one(model, text, amp):
    dev = next(model.parameters()).device
    b = tok(text, truncation=True, max_length=MAX_LEN, return_tensors='pt')
    with torch.inference_mode(), torch.autocast(dev.type, dtype=torch.float16, enabled=amp):
        lo = model(input_ids=b['input_ids'].to(dev), attention_mask=b['attention_mask'].to(dev)).logits
    return int((torch.log_softmax(lo.float(), -1).cpu().numpy()[0] + BIAS_F1).argmax())

def time_single(model, texts, amp, warm=10):
    for t in texts[:warm]:
        label_one(model, t, amp)
    ms = []
    for t in texts:
        t0 = time.perf_counter()
        label_one(model, t, amp)
        ms.append((time.perf_counter() - t0) * 1000)
    return np.array(ms)

lat_idx = np.random.default_rng(SEED + 9).choice(len(test_df), 20000, replace=False)
lat_txt = test_df['text'].to_numpy()[lat_idx]

m_gpu = AutoModelForSequenceClassification.from_pretrained(FT_DIR).to('cuda').eval()
_ = logp_batch(m_gpu, lat_txt[:512], 256, True)
torch.cuda.synchronize()
t0 = time.perf_counter()
lp_chk = logp_batch(m_gpu, lat_txt, 256, True)
torch.cuda.synchronize()
sec_gpu = time.perf_counter() - t0
agree = ((lp_chk + BIAS_F1).argmax(1) == np.asarray(PRED['bert'])[lat_idx]).mean()
ms_gpu = time_single(m_gpu, lat_txt[:300], True)

LAT = {'gpu_name': torch.cuda.get_device_name(0), 'n_batch': len(lat_txt), 'batch_size': 256,
       'gpu_batch_texts_per_sec': len(lat_txt) / sec_gpu,
       'gpu_single_ms_mean': float(ms_gpu.mean()), 'gpu_single_ms_p50': float(np.percentile(ms_gpu, 50)),
       'gpu_single_ms_p95': float(np.percentile(ms_gpu, 95)), 'gpu_agreement_with_saved_preds': float(agree)}
print(f'GPU: {LAT["gpu_name"]} | fp16')
print(f'daste-i (256 ta): {LAT["gpu_batch_texts_per_sec"]:.0f} nazar dar sanie ({len(lat_txt)} nazar dar {sec_gpu:.1f} s, ba tokenize)')
print(f'tak-nazar, az matn ta barchasb: mean {LAT["gpu_single_ms_mean"]:.1f} ms | p50 {LAT["gpu_single_ms_p50"]:.1f} | p95 {LAT["gpu_single_ms_p95"]:.1f}')
print(f'tavafogh ba pishbini-haye zakhire-shode: {agree:.4f}')
with open(f'{OUT}/latency.json', 'w') as f:
    json.dump(LAT, f, indent=1)
print('zakhire shod: latency.json')
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> همیشه GPU در دسترس نیست. باید دید مدل روی CPU چقدر کند است و آیا کوانتیزه کردن، بدون افت دقت، سریع‌ترش می‌کند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> مدل روی CPU دوهسته‌ای کولب یک بار با float32 و یک بار با کوانتیزه‌سازی دینامیک int8 لایه‌های Linear اجرا شد. هر دو روی همان ۲۰۰۰ نظر val مسیریاب سنجیده شدند. int8 فقط اگر کمتر از ۰٫۰۰۵ افت داشت پذیرفته می‌شد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> float32 برای هر نظر ۱۵۸ میلی‌ثانیه طول کشید و Macro F1 آن دقیقاً مثل GPU بود. int8 سرعت تک‌نظر را ۳٫۳۵ برابر کرد (۴۷ میلی‌ثانیه) و حجم مدل را از ۴۹۸ به ۲۴۱ مگابایت رساند، ولی Macro F1 را ۰٫۰۱۵۱ کم کرد و ۶٫۷ درصد پیش‌بینی‌ها را عوض کرد. پس پذیرفته نشد. حتی float32 روی همین CPU ساده حدود ۸ برابر از LLM سریع‌تر است.</p>

```python
import copy, tempfile, warnings
torch.set_num_threads(os.cpu_count() or 2)
m_cpu = AutoModelForSequenceClassification.from_pretrained(FT_DIR).float().eval()
with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    from torch.ao.quantization import quantize_dynamic
    m_q = quantize_dynamic(copy.deepcopy(m_cpu), {torch.nn.Linear}, dtype=torch.qint8).eval()
Q_METHOD = 'torch.ao.quantization.quantize_dynamic (Linear, qint8)'
print(f'CPU: {os.cpu_count()} hasteh | ravesh-e kvantize: {Q_METHOD}')


def size_mb(model):
    fd, path = tempfile.mkstemp(suffix='.pt')
    os.close(fd)
    torch.save(model.state_dict(), path)
    mb = os.path.getsize(path) / 1e6
    os.remove(path)
    return mb


acc_txt, acc_y = val_txt[r_idx], y_r
ref_pred = (lp_va_bert[r_idx] + BIAS_F1).argmax(1)
rows_c, preds_c = [], {}
for name, model in [('cpu_fp32', m_cpu), ('cpu_int8', m_q)]:
    t0 = time.perf_counter()
    lp_ = logp_batch(model, acc_txt, 64, False)
    sec = time.perf_counter() - t0
    preds_c[name] = (lp_ + BIAS_F1).argmax(1)
    ms = time_single(model, lat_txt[:200], False)
    rows_c.append({'setting': name, 'batch_texts_per_sec': len(acc_txt) / sec,
                   'single_ms_mean': ms.mean(), 'single_ms_p50': np.percentile(ms, 50),
                   'single_ms_p95': np.percentile(ms, 95), 'size_mb': size_mb(model),
                   'val_macro_f1': macro_f1(acc_y, preds_c[name]), 'agree_with_gpu': (preds_c[name] == ref_pred).mean()})
pred_32, pred_q = preds_c['cpu_fp32'], preds_c['cpu_int8']
CPU_DF = pd.DataFrame(rows_c).set_index('setting')
print(f'\nroo 2000 nazar-e val (hamoon nemoone-ye masiryab), GPU = {macro_f1(acc_y, ref_pred):.4f}:')
display(CPU_DF.round(4))

IDXQ = np.random.default_rng(SEED + 11).integers(0, len(acc_y), (2000, len(acc_y)))
dq = np.array([macro_f1(acc_y[ix], pred_q[ix]) - macro_f1(acc_y[ix], pred_32[ix]) for ix in IDXQ])
drop = CPU_DF.loc['cpu_fp32', 'val_macro_f1'] - CPU_DF.loc['cpu_int8', 'val_macro_f1']
speed = CPU_DF.loc['cpu_fp32', 'single_ms_mean'] / CPU_DF.loc['cpu_int8', 'single_ms_mean']
Q_ADOPT = drop <= 0.005
print(f'int8 dar barabar-e fp32: ofte-ye Macro F1 = {drop:+.4f} (bazeh-ye ekhtelaf {np.percentile(dq, 2.5):+.4f} ta '
      f'{np.percentile(dq, 97.5):+.4f}) | tavafogh-e pishbini {(pred_q == pred_32).mean():.4f} | '
      f'{speed:.2f} barabar sari-tar | hajm {CPU_DF.loc["cpu_fp32", "size_mb"]:.0f} -> {CPU_DF.loc["cpu_int8", "size_mb"]:.0f} MB')
print('tasmim (ghanoon-e 0.005):', 'int8 baraye CPU paziroofte shod' if Q_ADOPT else 'int8 paziroofte nashod, fp32 mimoone')

LAT.update({'quant_method': Q_METHOD, 'int8_adopted_for_cpu': bool(Q_ADOPT), 'int8_f1_drop_val': float(drop),
            'cpu_threads': os.cpu_count(), 'cpu': CPU_DF.reset_index().to_dict('records')})
with open(f'{OUT}/latency.json', 'w') as f:
    json.dump(LAT, f, indent=1, default=float)
print('zakhire shod: latency.json')
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> در کوانتیزه کردن BERT معمولاً لایه‌ی آخر حساس‌تر است، چون خطای گرد کردن مستقیم روی امتیاز نهایی کلاس‌ها می‌نشیند.</p>

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> نتیجه دقیقاً مثل int8 کامل شد (۰٫۷۳۲۵ و همان ۹۳٫۳ درصد توافق) و فقط کمی کندتر بود. پس افت از خود encoder می‌آید، نه از لایه‌ی آخر. حدس من این است که مرزهای تصمیم با بایاس جابه‌جا شده‌اند و نظرهای زیادی نزدیک مرز <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">no_idea</span> و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;direction:ltr;unicode-bidi:isolate">recommended</span> هستند، پس تغییر کوچکی در امتیازها برچسبشان را عوض می‌کند. مدل نهایی float32 ماند.</p>

```python
m_q2 = copy.deepcopy(m_cpu)
base = getattr(m_q2, m_q2.base_model_prefix)
with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    base.encoder = quantize_dynamic(base.encoder, {torch.nn.Linear}, dtype=torch.qint8)
m_q2.eval()

t0 = time.perf_counter()
lp_q2 = logp_batch(m_q2, acc_txt, 64, False)
sec_q2 = time.perf_counter() - t0
pred_q2 = (lp_q2 + BIAS_F1).argmax(1)
ms_q2 = time_single(m_q2, lat_txt[:200], False)
CPU_DF.loc['cpu_int8_encoder'] = {
    'batch_texts_per_sec': len(acc_txt) / sec_q2, 'single_ms_mean': ms_q2.mean(),
    'single_ms_p50': np.percentile(ms_q2, 50), 'single_ms_p95': np.percentile(ms_q2, 95),
    'size_mb': size_mb(m_q2), 'val_macro_f1': macro_f1(acc_y, pred_q2), 'agree_with_gpu': (pred_q2 == ref_pred).mean()}
display(CPU_DF.round(4))

dq2 = np.array([macro_f1(acc_y[ix], pred_q2[ix]) - macro_f1(acc_y[ix], pred_32[ix]) for ix in IDXQ])
drop2 = CPU_DF.loc['cpu_fp32', 'val_macro_f1'] - CPU_DF.loc['cpu_int8_encoder', 'val_macro_f1']
speed2 = CPU_DF.loc['cpu_fp32', 'single_ms_mean'] / CPU_DF.loc['cpu_int8_encoder', 'single_ms_mean']
Q2_ADOPT = drop2 <= 0.005
print(f'int8 (faghat encoder) dar barabar-e fp32: ofte-ye Macro F1 = {drop2:+.4f} (bazeh-ye ekhtelaf '
      f'{np.percentile(dq2, 2.5):+.4f} ta {np.percentile(dq2, 97.5):+.4f}) | tavafogh-e pishbini {(pred_q2 == pred_32).mean():.4f} | '
      f'{speed2:.2f} barabar sari-tar')
print('tasmim (ghanoon-e 0.005):', 'int8-e encoder baraye CPU paziroofte shod' if Q2_ADOPT else 'paziroofte nashod, fp32 mimoone')

LAT.update({'int8_encoder_only_adopted_for_cpu': bool(Q2_ADOPT), 'int8_encoder_only_f1_drop_val': float(drop2),
            'cpu': CPU_DF.reset_index().to_dict('records')})
with open(f'{OUT}/latency.json', 'w') as f:
    json.dump(LAT, f, indent=1, default=float)
print('zakhire shod: latency.json')
```
